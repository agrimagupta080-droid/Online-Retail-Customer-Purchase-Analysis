```python
#==========================================
# Project 21: Online Retail Customer Purchase Analysis
# Google Colab Execution Script (Fixed Column Mapping)
# ==========================================

# 1. Import Required Libraries
import pandas as pd
import numpy as np

import os
from google.colab import files

print("==================================================")
print("PROJECT : ONLINE RETAIL")
print("==================================================")

df = pd.read_csv("/content/online_retail.csv")

df.head()
```

```python
# 2. Load the CSV File and Inspect Dimensions, Column Names, and Data Types

print("\n--- 1. Dataset Shape & Info ---")
print(f"Dimensions (Rows, Columns): {df.shape}")
print("\nColumn Names and Data Types:")
print(df.dtypes)
print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ==========================================
# Column Mapping (Adapts standard CSV columns to assignment schema)
# ==========================================
column_mapping = {
    'InvoiceNo': 'Order_ID',
    'StockCode': 'Product_ID',
    'Description': 'Product_Name',
    'InvoiceDate': 'Order_Date',
    'UnitPrice': 'Unit_Price',
    'CustomerID': 'Customer_ID',
    'Country': 'Product_Category'  # Mapping Country as category if no category column exists
}
df = df.rename(columns=column_mapping)
```

```python
# ==========================================
# 3. Identify Missing Values and Duplicates
# ==========================================
print("\n--- 2. Missing Values & Duplicates ---")
print("Missing values per column:")
print(df.isnull().sum())

duplicate_orders = df.duplicated(subset=['Order_ID']).sum() if 'Order_ID' in df.columns else 0
print(f"Number of duplicate Order_ID records: {duplicate_orders}")
```

```python
# ==========================================
# 4. Handle Missing Quantity or Unit_Price Values
# ==========================================
initial_rows = len(df)
df = df.dropna(subset=['Quantity', 'Unit_Price'])
print(f"\n--- 3. Data Cleaning ---")
print(f"Dropped {initial_rows - len(df)} rows due to missing Quantity or Unit_Price.")
```

```python
# ==========================================
# 5. Convert Order_Date and Extract Date Features
# ==========================================
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df = df.dropna(subset=['Order_Date']) # Drop rows with invalid dates if any

df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Month_Name'] = df['Order_Date'].dt.month_name()
df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
# ==========================================
# 6. Calculate Total Orders, Units Sold, and Sales Revenue
# ==========================================
df['Order_Value'] = df['Quantity'] * df['Unit_Price']

total_orders = df['Order_ID'].nunique() if 'Order_ID' in df.columns else len(df)
total_units_sold = df['Quantity'].sum()
total_sales_revenue = df['Order_Value'].sum()

print("\n--- 5. Overall Business Metrics ---")
print(f"Total Orders: {total_orders:,}")
print(f"Total Units Sold: {total_units_sold:,}")
print(f"Total Sales Revenue: ${total_sales_revenue:,.2f}")
```

```python
# ==========================================
# 7. Find Top 10 Products with Highest Sales Revenue
# ==========================================
prod_col = 'Product_Name' if 'Product_Name' in df.columns else df.columns[1]
top_10_products = df.groupby(prod_col)['Order_Value'].sum().reset_index()
top_10_products = top_10_products.sort_values(by='Order_Value', ascending=False).head(10)

print("\n--- 7. Top 10 Products by Sales Revenue ---")
display(top_10_products)
```

```python
# ==========================================
# 8. Rank Product Categories by Quantity Sold and Revenue
# ==========================================
cat_col = 'Product_Category' if 'Product_Category' in df.columns else 'Country'
category_summary = df.groupby(cat_col).agg(
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('Order_Value', 'sum')
).reset_index()

category_summary['Revenue_Rank'] = category_summary['Total_Revenue'].rank(ascending=False, method='min')
category_summary['Quantity_Rank'] = category_summary['Total_Quantity'].rank(ascending=False, method='min')
category_summary = category_summary.sort_values(by='Total_Revenue', ascending=False)

print("\n--- 8. Category Rankings ---")
display(category_summary)
```

```python
# ==========================================
# 9. Compare Average Order Value Across Payment Methods
# ==========================================
if 'Payment_Method' in df.columns:
    payment_comparison = df.groupby('Payment_Method')['Order_Value'].mean().reset_index()
    payment_comparison.rename(columns={'Order_Value': 'Average_Order_Value'}, inplace=True)
    print("\n--- 9. Average Order Value by Payment Method ---")
    display(payment_comparison)
else:
    print("\n--- 9. Payment_Method column not present in dataset. Skipping. ---")
```

```python
# ==========================================
# 10. Identify Busiest Shopping Days and Peak Sales Month
# ==========================================
busiest_days = df.groupby('Weekday')['Quantity'].sum().reset_index().sort_values(by='Quantity', ascending=False)
peak_month = df.groupby(['Year', 'Month_Name'])['Order_Value'].sum().reset_index().sort_values(by='Order_Value', ascending=False).head(1)

print("\n--- 10. Busiest Days & Peak Month ---")
print("Busiest Shopping Days (by units sold):")
display(busiest_days)
print("\nPeak Sales Month:")
display(peak_month)
```

```python
# ==========================================
# 11. Dictionary for Category Sales Summary & Set for Unique Categories
# ==========================================
category_sales_dict = pd.Series(category_summary['Total_Revenue'].values, index=category_summary[cat_col]).to_dict()
unique_categories = set(df[cat_col].dropna().unique())

print("\n--- 11. Dictionary & Set Examples ---")
print(f"Unique Categories Set: {list(unique_categories)[:5]}... (Total: {len(unique_categories)})")
```

```python
# ==========================================
# 12. Function to Get Total Sales by Category Name
# ==========================================
def get_category_sales(category_name):
    """Returns total sales for a given product category."""
    if category_name in category_sales_dict:
        return category_sales_dict[category_name]
    return 0.0

sample_cat = list(unique_categories)[0] if unique_categories else "N/A"
print(f"\n--- 12. Function Test ---")
print(f"Total sales for category '{sample_cat}': ${get_category_sales(sample_cat):,.2f}")
```

```python
# ==========================================
# 13. NumPy Statistics for Order_Value
# ==========================================
order_values_arr = df['Order_Value'].to_numpy()

mean_val = np.mean(order_values_arr)
median_val = np.median(order_values_arr)
std_val = np.std(order_values_arr)

print("\n--- 13. NumPy Order Value Statistics ---")
print(f"Mean Order Value: ${mean_val:,.2f}")
print(f"Median Order Value: ${median_val:,.2f}")
print(f"Standard Deviation: ${std_val:,.2f}")
```

```python
# ==========================================
# 14. IQR Method for Outliers & Order-Value Bands
# ==========================================
Q1 = np.percentile(order_values_arr, 25)
Q3 = np.percentile(order_values_arr, 75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

outliers = df[df['Order_Value'] > upper_bound]
print(f"\n--- 14. IQR & Outliers ---")
print(f"IQR: {IQR:,.2f} | Upper Bound Threshold: ${upper_bound:,.2f}")
print(f"Number of unusually high order values (outliers): {len(outliers):,}")

low_thresh = np.percentile(order_values_arr, 33)
high_thresh = np.percentile(order_values_arr, 66)

def assign_value_band(val):
    if val <= low_thresh:
        return 'Low'
    elif val <= high_thresh:
        return 'Medium'
    else:
        return 'High'

df['Order_Value_Band'] = df['Order_Value'].apply(assign_value_band)
print("\nOrder Value Band Distribution:")
print(df['Order_Value_Band'].value_counts())
```

```python
# ==========================================
# 15. Product Class & Recommendations
# ==========================================
class Product:
    def __init__(self, name, price, category=None):
        self.name = name
        self.price = price
        self.category = category

    def display_info(self):
        return f"Product: {self.name} | Price: ${self.price:.2f}"

print("\n--- 15. Product Class Instantiation Example ---")
sample_product = Product(name="Sample Top Item", price=45.99, category=sample_cat)
print(sample_product.display_info())

# Save cleaned dataset
df.to_csv('cleaned_online_retail.csv', index=False)
print("\nCleaned dataset saved successfully as 'cleaned_online_retail.csv'.")
```
